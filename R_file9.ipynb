# Phase 3.1 — GDSC labels 

```python

# Keep only required columns from GDSC breast response
gdsc_br = gdsc_breast[[
    "Drug Name",
    "Cell Line Name",
    "AUC"
]].copy()

print(gdsc_br.shape)
gdsc_br.head()
```

```python
# assign breast cancer subtypes to cell lines
```

```python

# Curated breast cancer cell line subtype mapping (standard)

subtype_map = {
    # Luminal
    "MCF7": "LUMA",
    "T47D": "LUMA",
    "ZR-75-1": "LUMA",
    "BT-483": "LUMA",
    "BT-474": "LUMB",
    "HCC1428": "LUMB",

    # HER2+
    "AU565": "HER2",
    "SK-BR-3": "HER2",
    "HCC1954": "HER2",

    # TNBC
    "BT-20": "TNBC",
    "BT-549": "TNBC",
    "HCC1143": "TNBC",
    "HCC1187": "TNBC",
    "HCC1395": "TNBC",
    "HCC1599": "TNBC",
    "MDA-MB-231": "TNBC",
    "MDA-MB-468": "TNBC"
}

# Map subtypes
gdsc_br["subtype"] = gdsc_br["Cell Line Name"].map(subtype_map)

# Drop unmapped cell lines
gdsc_br = gdsc_br.dropna(subset=["subtype"])

print(gdsc_br["subtype"].value_counts())
gdsc_br.head()
```

```python
# **convert cell-line–level AUC → drug × subtype evidence**.


# Aggregate AUC to drug × subtype level
drug_subtype_auc = (
    gdsc_br
    .groupby(["Drug Name", "subtype"])["AUC"]
    .median()
    .reset_index()
)

print(drug_subtype_auc.shape)
drug_subtype_auc.head()
```

```python
# **AUC → binary labels** using your predefined rules.

'''
- AUC < **0.2** → **Sensitive (label = 1)**
- AUC > **0.8** → **Resistant (label = 0)**
- Else → **Uncertain (drop for now)**
'''

def auc_to_label(auc):
    if auc < 0.2:
        return 1
    elif auc > 0.8:
        return 0
    else:
        return np.nan  # uncertain

drug_subtype_auc["label"] = drug_subtype_auc["AUC"].apply(auc_to_label)

# Keep only confident labels
gdsc_labels = drug_subtype_auc.dropna(subset=["label"]).copy()

gdsc_labels["label"] = gdsc_labels["label"].astype(int)

print(gdsc_labels.shape)
gdsc_labels["label"].value_counts()
```

```python
## ✅ Using relaxed rule (Option B)

'''
- **label = 1 (sensitive)** → AUC < **0.5**
- **label = 0 (resistant)** → AUC > **0.8**
- Else → drop (uncertain)
'''


# Recompute labels with relaxed threshold
def auc_to_label_relaxed(auc):
    if auc < 0.5:
        return 1
    elif auc > 0.8:
        return 0
    else:
        return np.nan

drug_subtype_auc["label"] = drug_subtype_auc["AUC"].apply(auc_to_label_relaxed)

gdsc_labels = drug_subtype_auc.dropna(subset=["label"]).copy()
gdsc_labels["label"] = gdsc_labels["label"].astype(int)

print(gdsc_labels.shape)
gdsc_labels["label"].value_counts()
```

```python
### Add confidence & evidence source


gdsc_labels["confidence"] = "medium"
gdsc_labels["evidence"] = "GDSC"

gdsc_labels.head()
```

```python
import pandas as pd

ct = pd.read_csv(
    r"C:\Users\deep8\breast_cancer_project_folder\ctg-studies.csv"
)

print(ct.shape)
ct.head()
```

```python
### Extract drugs from trials

# Extract drug names from Intervention column
ct_drugs = (
    ct[["NCT Number", "Interventions", "Phases", "Study Status", "Conditions"]]
    .dropna(subset=["Interventions"])
    .copy()
)

ct_drugs.head()
```

```python

import re

def extract_drugs(intervention_str):
    # Remove prefixes like DRUG:, BIOLOGICAL:, etc.
    parts = intervention_str.split("|")
    drugs = []
    for p in parts:
        p = re.sub(r"^[A-Z_ ]+:\s*", "", p)
        drugs.append(p.strip().lower())
    return drugs

ct_drugs["drug_list"] = ct_drugs["Interventions"].apply(extract_drugs)

ct_drugs[["Interventions", "drug_list"]].head()
```

```python

# Detect subtype from Conditions text
def detect_subtype(cond):
    c = cond.lower()
    if "her2" in c:
        return "HER2"
    elif "triple negative" in c or "tnbc" in c:
        return "TNBC"
    elif "luminal" in c or "er+" in c or "estrogen receptor" in c:
        return "LUMA"
    else:
        return "PAN"  # non-specific breast cancer

ct_drugs["subtype"] = ct_drugs["Conditions"].apply(detect_subtype)

ct_drugs[["Conditions", "subtype"]].head()
```

```python

# Clinical label assignment
def clinical_label(row):
 if row["Study Status"] in ["COMPLETED", "ACTIVE, NOT RECRUITING", "RECRUITING"]:
     if row["Phases"] in ["PHASE2", "PHASE3"]:
         return 1
 return 0

ct_drugs["clinical_label"] = ct_drugs.apply(clinical_label, axis=1)

ct_drugs[["Phases", "Study Status", "clinical_label"]].value_counts()
```

```python

### Explode drug lists (one drug per row)

ct_exploded = ct_drugs.explode("drug_list").copy()

# Keep relevant columns
ct_exploded = ct_exploded[[
    "NCT Number",
    "drug_list",
    "subtype",
    "clinical_label",
    "Phases",
    "Study Status"
]]

ct_exploded.rename(columns={"drug_list": "drug_name"}, inplace=True)

print(ct_exploded.shape)
ct_exploded.head()
```

```python

### Normalize drug names & drop non-therapeutics

def clean_drug_name(name):
    name = name.lower()
    name = name.replace(" hydrochloride", "")
    name = name.replace(" dp", "")
    name = name.replace("-", " ")
    return name.strip()

ct_exploded["drug_name_clean"] = ct_exploded["drug_name"].apply(clean_drug_name)

# Remove placebo / supplements
remove_terms = ["placebo", "acetyl", "vitamin", "supplement"]
ct_exploded = ct_exploded[
    ~ct_exploded["drug_name_clean"].str.contains("|".join(remove_terms))
].copy()

print(ct_exploded.shape)
ct_exploded[["drug_name", "drug_name_clean"]].head()
```
